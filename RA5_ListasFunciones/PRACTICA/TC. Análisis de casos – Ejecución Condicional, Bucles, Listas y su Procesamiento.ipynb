# Caso 1: Control de ventas diarias de una soda

## Problema
Una soda ubicada en San Pedro registra sus ventas de lunes a viernes. El programa debe calcular el total semanal, el promedio diario y clasificar el resultado según una meta.

## Entrada / datos
Use esta lista base:

```python
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
```

## Proceso esperado

- Calcular total semanal.
- Calcular promedio diario.
- Comparar el total con la meta.
- Clasificar:
  - `Meta alcanzada` si el total es mayor o igual a la meta.
  - `Cerca de la meta` si el total alcanza al menos el 80% de la meta.
  - `Requiere atención` si no alcanza el 80%.

## Salida esperada
Debe mostrar total, promedio, porcentaje de cumplimiento y estado final.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[85000, 92000, 78000, 110000, 97000]`, meta `450000` | Meta alcanzada | Meta Alcanzada |
| Caso límite | total exactamente igual a la meta | Meta alcanzada | Meta alcanzada |
| Caso especial | total menor al 80% de la meta | Requiere atención | requiere atencion |

````python
# Caso 1 - Escriba su solución aquí

ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
#Calcular total semanal

def calcular_total(lista_ventas):
    total = 0
    
    for venta in lista_ventas:
        total += venta
        
    return total
total_semanal = calcular_total(ventas_soda)
print("total semanal:", total_semanal)
#Calcular promedio diario
def calcular_promedio(total, cantidad):
    promedio = total / cantidad
    return promedio
promedio = calcular_promedio(total_semanal, len(ventas_soda))
print("promedio diario:",int(promedio))
#Comparar el total con la meta

listas_ventas = [
    [85000, 92000, 78000, 110000, 97000],   # Meta alcanzada
    [70000, 75000, 72000, 80000, 70000],    # Cerca de la meta
    [50000, 60000, 55000, 65000, 60000]     # Requiere atención
]

for ventas in listas_ventas:

    total = sum(ventas)

    print("Total:", total)

    if total >= meta_semanal:
        print("Meta alcanzada")

    elif total >= meta_semanal * 0.80:
        print("Cerca de la meta")

    else:
        print("Requiere atención")

    print("--------------------")
#Clasificar: Meta alcanzada si es mayor o igual a la meta.
#Cerca la meta si el total alcanza al menos 80% de la meta 
#Requiere atencion si no alcanza el 80%
# Complete el análisis
````

# Caso 2: Asistencia a una actividad cultural

## Problema
Una actividad cultural en Costa Rica registra asistencia por bloques horarios. El programa debe calcular la asistencia promedio y clasificar el nivel de ocupación.

## Entrada / datos
Use esta lista base:

```python
asistencias = [120, 180, 210, 250, 230]
capacidad = 300
```

## Proceso esperado

- Recorrer la lista con un ciclo `for`.
- Calcular asistencia total y promedio.
- Contar cuántos bloques superaron el 70% de capacidad.
- Clasificar el promedio de ocupación:
  - `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad.
  - `Ocupación media` si el promedio es mayor o igual al 50%.
  - `Baja ocupación` si el promedio es menor al 50%.

## Salida esperada
Debe mostrar total de asistentes, promedio, bloques sobre 70% y estado de ocupación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[120, 180, 210, 250, 230]`, capacidad `300` | Ocupación media o alta según cálculo | |
| Caso límite | promedio exactamente 80% de capacidad | Alta ocupación | |
| Caso especial | lista con asistencias muy bajas | Baja ocupación | |

````python
# Caso 2 - Escriba su solución aquí

asistencias = [120, 180, 210, 250, 230]
capacidad = 300

#Recorrer la lista en ciclo for
def mostrar_asistencias(lista):
    for asistencia in lista:
        print(asistencia)
        
mostrar_asistencias(asistencias)
print("----------------------")
#Calcular asistencia total y promedio
def calcular_total(lista):
    return sum(lista)

def calcular_promedio(lista):
    return sum(lista) / len(lista)

total = calcular_total(asistencias)
promedio = int(calcular_promedio(asistencias))

print("Asistencia Total:", total)
print("Promedio:", promedio)
#Contar cuantos bloques superaron el 70% de capacidad
print("----------------------------")
def contar_bloques(lista, capacidad):

    contador = 0
    for asistencia in lista:
        if asistencia > capacidad * 0.70:
            contador += 1
    return contador
cantidad = contar_bloques(asistencias, capacidad)

print("Cuanta cantidad superal el 70 porciento de capacidad?: ",cantidad)
#CLasificar el promedio de ocupacion: `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad. 
def clasificar_ocupacion(promedio, capacidad):

    if promedio >= capacidad * 0.80:
        return "Alta ocupación"

    elif promedio >= capacidad * 0.50:
        return "Ocupación media"

    else:
        return "Baja ocupación"
clasificacion = clasificar_ocupacion(promedio, capacidad)

print("promedio de ocupacion: ", clasificacion)

# - `Ocupación media` si el promedio es mayor o igual al 50%. 

# - `Baja ocupación` si el promedio es menor al 50%.
````

# Caso 3: Precios de productos de feria

## Problema
Una persona compara precios de un mismo producto en diferentes ferias del agricultor. El programa debe analizar los precios y detectar si hay mucha variación.

## Entrada / datos
Use esta lista base:

```python
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]
```

## Proceso esperado

- Calcular precio promedio.
- Identificar precio mínimo y máximo.
- Calcular rango de precios: `máximo - mínimo`.
- Clasificar la variación:
  - `Variación alta` si el rango es mayor o igual a 300.
  - `Variación moderada` si el rango es mayor o igual a 150.
  - `Variación baja` si el rango es menor a 150.
- Usar un ciclo para contar cuántos precios están por encima del promedio.

## Salida esperada
Debe mostrar producto, promedio, mínimo, máximo, rango, precios sobre promedio y clasificación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[900, 850, 1000, 950, 1200]` | Variación alta | |
| Caso límite | rango exactamente `300` | Variación alta | |
| Caso especial | todos los precios iguales | Variación baja | |

````python
# Caso 3 - Escriba su solución aquí

producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]

total = sum(precios)

promedio = total / 5
print(f"promedio: {promedio}")

minimo = min(precios)
print(f"promedio minimo: {minimo}")

maximo = max(precios)
print(f"precio minimo: {maximo}")

rango = maximo - maximo
print(f"el rango de precios es: {rango}")

if rango >= 300:
    print("variacion alta")
elif rango >= 150:
    print("varicion moderada")
else:
    print("variacion")
````
